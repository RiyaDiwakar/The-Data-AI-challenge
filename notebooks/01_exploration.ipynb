```python
import json

candidate_lookup = {}

with open(
    "../data/candidates.jsonl",
    "r",
    encoding="utf-8"
) as f:

    for line in f:
        candidate = json.loads(line)
        candidate_lookup[candidate["candidate_id"]] = candidate

print("Loaded:", len(candidate_lookup))
```

```python
import pandas as pd

def generate_reasoning(candidate):

    title = candidate["profile"].get(
        "current_title",
        "Professional"
    )

    summary = candidate["profile"].get(
        "summary",
        ""
    )

    summary_lower = summary.lower()

    strengths = []

    if "retrieval" in summary_lower:
        strengths.append("retrieval systems")

    if "ranking" in summary_lower:
        strengths.append("ranking models")

    if "search" in summary_lower:
        strengths.append("search relevance")

    if "recommendation" in summary_lower:
        strengths.append("recommendation systems")

    if "embedding" in summary_lower:
        strengths.append("embedding-based search")

    if "machine learning" in summary_lower:
        strengths.append("machine learning")

    if len(strengths) == 0:
        strengths.append("AI/ML expertise")

    strengths_text = ", ".join(strengths[:4])

    return (
        f"{title} demonstrates strong alignment with the role through "
        f"experience in {strengths_text}. "
        f"The profile highlights relevant production AI/ML work and shows "
        f"capabilities closely matching the job requirements."
    )


submission_rows = []

for rank, r in enumerate(results[:100], start=1):

    candidate = candidate_lookup[
        r["candidate_id"]
    ]

    submission_rows.append({
        "candidate_id": r["candidate_id"],
        "rank": rank,
        "score": float(r["score"]),
        "reasoning": generate_reasoning(candidate)
    })

submission = pd.DataFrame(
    submission_rows
)

submission.to_csv(
    "../outputs/final_submission.csv",
    index=False
)

print(submission.head())
print("Rows:", len(submission))
```

```python
def generate_reasoning(candidate):

    title = candidate["profile"].get("current_title", "")
    summary = candidate["profile"].get("summary", "").lower()

    reasons = []

    keywords = [
        "search",
        "retrieval",
        "ranking",
        "recommendation",
        "nlp",
        "embedding",
        "machine learning"
    ]

    for k in keywords:
        if k in summary:
            reasons.append(k)

    if reasons:
        return (
            f"{title} demonstrates strong alignment with the role through experience in "
            f"{', '.join(reasons[:4])}. Profile shows relevant production AI/ML expertise."
        )

    return (
        f"{title} demonstrates strong alignment with the Senior AI Engineer role based on overall AI/ML experience and semantic similarity to the job requirements."
    )
```
